```python
'''renaming columns'''
import pandas as pd
df=pd.read_csv('data.csv')
df=df.rename(columns={'Date':'sales__date'})
df.head()
```

```python
###changing data types
df.dtypes
df['NEW_value']=df['Value'].fillna(df['Value'].mean()).astype(int)
print(df)



```

```python
df['modified_value']=df['Value'].apply(lambda x:x*2)
print(df)
```

```python
##Data aggregating and grouping
df.head()
```

```python
grouped_mean=df.groupby('Category')['Value'].mean()
print(grouped_mean)
```

```python
df.groupby(['Category','Region'])['Value'].sum()
```

```python
##aggregate multiple functions
grouped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
###merging and joining dataframes
df1=pd.DataFrame({'key':['A','B','C'],'value1':[1,2,3]})
df2=pd.DataFrame({'key':['A','B','D'],'value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
''''merging of two dataframes on key column'''
pd.merge(df1, df2, how='inner', on='key')
```

```python
pd.merge(df1,df2,how='outer',on='key')
```

```python
pd.merge(df1,df2,how='left',on='key')
```

```python
pd.merge(df1,df2,how='right',on='key')###right dataframe will be given priority
```
