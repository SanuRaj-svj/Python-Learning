unordered collection of data
stored data in form of key-value pairs
keys must be unique and immutable(eg.strings,numbers,or tuples),while values can be of any type.

```python
##creating dictionaries
empty_dict={}
print(type(empty_dict))
##or
empty_dict=dict()
print(type (empty_dict))
```

```python
student={"name":"sanu","age":20,"course":"data-science"}
print(type(student))
print(student)
```

```python
student={"name":"sanu","age":20,"course":"data-science","course":"machine-learning"}
print(student)
```

```python
##accessing dictionary elements
student={"name":"sanu","age":20,"course":"data-science"}
print(student["name"])
print(student["age"])
print(student["course"])
##using get()method
print(student.get("name"))
print(student.get("grade"))
print(student.get("grade","not found"))
```

```python
###modifying dictionary elements
###dictionary are mutable so we can add,update or delete elements
print(student)
```

```python
student["age"]=21###updating value
print(student)
student["grade"]="A"###adding new key with value
print(student)
```

```python
del student["grade"]###deleting key with values
print(student)
```

```python
###Dictionary methods
keys=student.keys()##get all keys
print(keys)
values=student.values()##get all values
print(values)
items=student.items()##get all key-value pairs
print(items)
```

```python
student_copy=student
print(student_copy)
print(student)
```

```python
student["age"]=22
print(student_copy)
print(student)
```

```python
student_copy1=student.copy()
print(student_copy1)
print(student)
```

```python
###shallow copy of dictionary
student["age"]=23
print(student_copy1)
print(student)
```

```python
###iteration over keys in dictionary
for key in student.keys():
  print(key)
```

```python
###iteration over values in dictionary
for value in student.values():
  print(value)
```

```python
###iteration over both key value pairs
for key,value in student.items():
  print(f"{key}:{value}")
```

```python
###nested Dictionaries

student={
  "student1":{"name":"sanu","age":20},
  "student2":{"name":"raj","age":21}
}
print(student)
```

```python
###accessing items from nested dictionaries
###use keys only to deal with it
print(student["student1"]["name"])
print(student["student2"]["age"])
```

```python
###iteration over nested dictionaries
for student_id,student_info in student.items():
  print(f"{student_id}:{student_info}")
  for key,value in student_info.items():
    print(f"{key}:{value}")
```

```python
###dictionaries comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
###with condition
even_squares={x:x**2 for x in range(5) if x%2==0}
print(even_squares)
```

```python
###without using dictionaries comprehension
for x in range(5):
  if x%2==0:

    print(f"{x}:{x**2}")
```

```python
###use a dictionary to count  frequency of elements in list
numbers=[1,2,2,3,4,4,4,4]
frequency={}
for number in numbers:
  if number in frequency:
    frequency[number]+=1
  else:
      frequency[number]=1
print(frequency)
  
```

```python
###merging two dictionaries
dict_1={"a":1,"b":2}
dict_2={"c":3,"d":4}
merged_dict={**dict_1,**dict_2}
print(merged_dict)
```
