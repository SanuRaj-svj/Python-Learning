##tuples
ordered collections of items that are immutable. they are similar to lists but their immutability makes them different

```python
##creating tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
##other way to create tuple
tup=tuple()
print(type(tup))
```

```python
###type conversion
###list to tuple 
numbers=tuple([1,2,3,4,5])
print(numbers)
##tuple to list
numbers=list((1,2,3,4,5))
print(numbers)
numbers=tuple([1,2,3,4,5])
###mixed tuple
mixed_tuple=(1,'hello world',True)
print(mixed_tuple)
```

```python
print(numbers)
mixed_tuple
```

```python
print(numbers[::])
print(numbers[0:3])
print(numbers[:-1])
print(numbers[::-1])
print(numbers[0:3:2])
```

```python
concat_tuple=numbers+mixed_tuple
print(concat_tuple)
print(concat_tuple*3)
print(numbers*3)
```

```python
##immutable nature of tuple
##once defined elements cannot be changed
lst=[1,2,3,4,5]
lst[1]="sanu"
print(lst)
tuple=(1,2,3,4,5)
tuple[1]="sanu"
print(tuple)
```

##way to assign item in tuple is that firstly we convert tuple into list and then assign the change and again convert the list into tuple with changed item

```python
##tuple methods
print(numbers.count(1))
print(numbers.index(3))
```

```python
##packing and unpacking tuple
##packing
packed_tuple=1,"hello",3.14
print(packed_tuple)
##unpacking
a,b,c=packed_tuple
print(a)
print(b)
print(c)
```

```python
##unpacking with star
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
##nested list
lst=[[1,2,3],[4,5,6],[7,8,9]]
print(lst[0][0])
print(lst[0][0:2])
##tuple inside list
lst=[[1,2,3],[4,5,6],(7,8,"hello")]
print(lst[2][2])


###nested tuple
tup=((1,2,3),(4,5,6),(7,8,9))
print(tup[0][0])
```
