```python
##create a set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_set=set((1,2,3))
print(my_set)
my_set=set([1,2,3])
print(my_set)
```

```python
my_set={1,2,3,4,5,6,7,5,6,8}
print(my_set)
```

```python
##basic set operations
##adding and removing elements
my_set.add(9)
print(my_set)
my_set.add(9)##ignore
print(my_set)
```

```python
my_set.remove(3)
print(my_set)
my_set.remove(10)
print(my_set)
```

```python
my_set.discard(10)##if 10 is not in set then it will print set without error used as alternative of remove
print(my_set)
```

```python
removed_element=my_set.pop()##first in first out in set
print(removed_element)
print(my_set)
```

```python
my_set.clear()
print(my_set)
```

```python
###membership test in set

my_set={1,2,3,4,4,5}
print(3 in my_set)
print(4 in my_set)
print(8 in my_set)
```

```python
###uniion of two sets
set1={1,2,3}
set2={3,2,4,5,6}
union_set=set1.union(set2)
print(union_set)
###intersection of two sets
intersection_set=set1.intersection(set2)
print(intersection_set)
set1.intersection_update(set2)
print(set1)
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}
print(set1.difference(set2))
print(set1)
print(set2.difference(set1))
print(set1.difference_update(set2))
print(set1)
```

```python
set1={1,2,3,4,5}
set2={4,5,6,7,8}
print(set1.symmetric_difference(set2))
```

```python
###sets methods
set1={1,2,3,4,5}
set2={3,4,5}
##is subset
print(set1.issubset(set2))
print(set1.issuperset(set2))
```

```python
###counting unique words in text
text="in this tutorial we are going to learn sets in python,are we ready"
words=text.split()
unique_words=set(words)
print(unique_words)
print(len(unique_words))
```
