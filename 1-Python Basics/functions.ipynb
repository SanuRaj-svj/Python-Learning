```python
###basic syntax
def even_or_odd(num):
  """this function finds number is even or odd"""
  if num%2==0:
    print("the number is even")
  else:
      print("the number is odd")
even_or_odd(2)
```

```python
###multiple parameters functions
def add(a,b):
  c=a+b
  return c
add(2,3)
```

```python
even_or_odd(24)
```

```python
result=add(1,2)
print(result)
```

```python
###default parameter
def greet(name="guest"):
  print(f"hello {name} welcome to paradise")
greet()
greet("sanu")
 
```

```python
###variable length arguments
##positional argument
def print_numbers(*args):
  for numbers in args:
    print(numbers)
```

```python
###keyword arguments
def print_details(**kwargs):
  for key,value in kwargs.items():
    print(f"{key}:{value}")
```

```python
print_details(name="sanu",age="20",city="bhopal")
```

```python
print_numbers(1,2,3,4,5,6,7,8,"sanu")
```

```python
def print_arguments(*args,**kwargs):
  for arg in args:
    print(f" positional argument:{arg}")
  for key,value in kwargs.items():
    print(f"{key}:{value}")  
```

```python
print_arguments(1,2,3,name="sanu",age="20")
```
