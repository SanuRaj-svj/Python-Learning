```python
def convert_temperature(temp,unit):
  if unit=='c':
    return 9/5*temp+32
  elif unit=='f':
    return 5/9*(temp-32)
  else:  
    return "Invalid unit"
```

```python
convert_temperature(100,'c')
```

```python
###check strong passwpord or not
def pwdcheck(password):
  if(len(password)<8):
    return "weak password"
  elif not any(char.isdigit() for char in password):
    return "weak password"
  elif not any(char.islower() for char in password):
      return "weak password"
  elif not any(char.isupper() for char in password):
      return "weak password"
  else:
        return "strong password"
```

```python
pwdcheck("Sanu1234")
```

```python
###check pallindrome
def is_pallindrome(s):
  s=s.lower().replace(" ","")
  return s==s[::-1]
```

```python
is_pallindrome("hello")
```
